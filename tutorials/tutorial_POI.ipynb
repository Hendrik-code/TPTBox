# POI
## Summary
This code defines the `POI` class, which represents a collection of centroids used to define points of interest in medical imaging data. It provides various methods for manipulating and working with the centroid data.

## Example Usage


````python
# Create a POI object with 2D dictionary input
from TPTBox.core.poi import POI

poi_data = {
    (1, 30): (10.0, 20.0, 30.0),
    (2, 1): (15.0, 25.0, 35.0),
}
poi_obj = POI(centroids=poi_data, orientation=("R", "A", "S"), zoom=(1.0, 1.0, 1.0), shape=(256, 256, 100))

# Access attributes
print(poi_obj.orientation)
print(poi_obj.centroids)
print(poi_obj.zoom)
print(poi_obj.shape)

# Update attributes
poi_obj.rescale_((2.0, 2.0, 2.0))
poi_obj.centroids[(3, 50)] = (5.0, 15.0, 25.0)
print(poi_obj)

# Create a copy of the object
poi_copy = poi_obj.copy()

# Perform operations
print(poi_obj)
poi_obj = poi_obj.map_labels(label_map_region={1: 4, 2: 8})
poi_obj.round_(0)
print(poi_obj)
````


## Code Analysis
### Main functionalities
- Represents a collection of centroids used to define points of interest in medical imaging data.
- Provides methods for manipulating and working with the centroid data.
___
### Methods
- `is_global`: Property indicating whether the POI is a global POI.
- `zoom`: Property getter for the zoom level.
- `affine`: Property representing the affine transformation for the image.
- `clone`: Creates a copy of the POI object with optional attribute overrides.
- `copy`: Creates a copy of the POI object with optional attribute overrides.
- `local_to_global`: Converts local coordinates to global coordinates using zoom, rotation, and origin.
- `global_to_local`: Converts global coordinates to local coordinates using zoom, rotation, and origin.
- `apply_crop`: Crops the centroids based on the given origin shift due to the image crop.
- `shift_all_coordinates`: Shifts all centroid coordinates based on the given translation vector.
- `reorient`: Reorients the centroids of an image from the current orientation to the specified orientation.
- `rescale`: Rescales the centroid coordinates to a new voxel spacing in the current x-y-z-orientation.
- `resample_from_to`: Resamples the centroids to the voxel mapping of another image / POI.
- `map_labels`: Maps regions and subregions to new regions and subregions based on a label map dictionary.
- `save`: Saves the centroids to a JSON file.
- `sort`: Sorts the centroids by the sorting list.
- `make_point_cloud_nii`: Creates point cloud NIfTI images from the centroid coordinates.
- `filter_points_inside_shape`: Filters out centroid points that are outside the defined shape.
- `round`: Rounds the centroid coordinates to a specified number of digits.
- `calculate_distances`: Calculates the distances between the target point and each centroid.
- `remove_centroid`: Removes centroid points with the specified labels.
___
### Fields
- `orientation`: The orientation of the image.
- `zoom`: The zoom level of the image.
- `shape`: The shape of the image.
- `sorting_list`: The order of the centroid points.
- `format`: The format of the image.
- `info`: Additional information stored as key-value pairs.
- `rotation`: The rotation matrix for the image orientation.
- `origin`: The origin of the image in millimeters along the x, y, and z axes.

# calc_poi_from_subreg_vert
## Summary
The `calc_poi_from_subreg_vert` function calculates the POIs of a subregion within a vertebral mask. It takes in a vertebral mask image reference and a subregion image reference, along with optional parameters such as the number of decimal places to round the output coordinates, the subregion ID(s) to calculate POIs for, and whether to print progress messages. If multiple subregion IDs are provided, the function computes all of them and aggregates the coordinates into a single `POI` object. The function returns a `POI` object containing the calculated POI coordinates.

## Example Usage
```python
# Calculate POIs for a subregion within a vertebral mask
vert_msk = NII.load(..., True)
subreg = NII.load(..., True)
centroids = calc_poi_from_subreg_vert(vert_msk, subreg, decimals=1, subreg_id=50)

# Print the coordinates
print(centroids.centroids)
```

## Code Analysis
### Inputs
- `vert` (Image_Reference): A vertebral mask image reference.
- `subreg` (Image_Reference): An image reference for the subregion of interest.
- `decimals` (int, optional): Number of decimal places to round the output coordinates to. Defaults to 2.
- `subreg_id` (int | Location | list[int | Location], optional): The ID(s) of the subregion(s) to calculate POIs for. Defaults to 50.
- `verbose` (bool, optional): Whether to print progress messages. Defaults to False.
- `extend_to` (POI | None, optional): An existing POI object to extend with the new POI values. Defaults to None.
- `buffer_file` (str | Path | None, keyword-only): Optional cache file. When provided and existing, the POIs are loaded from it and only missing points are (re-)computed.
- `save_buffer_file` (bool, keyword-only): If True, persist the resulting POI back to `buffer_file` when new points were added.
___
### Outputs
- `POI`: A `POI` object containing the calculated POI coordinates.
___


# calc_centroids
## Summary
The `calc_centroids` function calculates the centroid coordinates of each region in a given mask image.

## Example Usage
```python
msk = NII.load(..., True)
centroids = calc_centroids(msk, decimals=3, first_stage=-1, second_stage=50, extend_to=None)
```

## Code Analysis
### Inputs
- `msk` (Image_Reference): The input mask image.
- `decimals` (int, optional): The number of decimal places to round the centroid coordinates to. Default is 3.
- `first_stage` (int | Location, optional): The fixed value for the first (region) coordinate of each POI key. Use `-1` to take the value from the mask label instead. Default is `-1`.
- `second_stage` (int | Location, optional): The fixed value for the second (subregion) coordinate of each POI key. Use `-1` to take the value from the mask label instead. Default is `50`.
- `extend_to` (POI | None, optional): An existing Centroids object to add the calculated centroids to. Default is None.
- `inplace` (bool, optional): If True and `extend_to` is provided, mutate it in place instead of copying. Default is False.

Exactly one of `first_stage` / `second_stage` must be `-1`; the other is stored as a fixed value.
___
### Flow
1. Convert the input mask image to a NII object using the `to_nii` function.
2. Get the segmentation array from the NII object.
3. Get the affine transformation matrix from the NII object.
4. Create an empty `POI_Descriptor` object to store the centroid coordinates.
5. Get the unique non-zero values from the segmentation array.
6. Remove any NaN values from the unique values.
7. Iterate over each unique value:
   - Create a temporary mask array for the current value.
   - Calculate the center of mass of the temporary mask array.
   - Round the centroid coordinates to the specified number of decimal places.
   - Store the centroid coordinates in the `POI_Descriptor` object based on the fixed dimensions.
8. If `extend_to` is None, create a new `POI` object with the centroid coordinates and return it.
   Otherwise, update the `extend_to` object with the calculated centroid coordinates and return it.
___
### Outputs
- `POI` object: A `POI` object containing the calculated centroid coordinates.
___


# calc_poi_labeled_buffered
## Summary
`calc_poi_labeled_buffered` (alias for `calc_poi_from_two_segs`) computes the centroids of a given mask with respect to a subregion mask. It saves the centroids to a file or loads them from an existing file if it already exists.

## Example Usage
```python
# Example 1: Compute centroids and save them to a file
msk_reference = NII.load(..., True)
subreg_reference = NII.load(..., True)
out_path = Path(...)
centroids = calc_poi_labeled_buffered(msk_reference, subreg_reference, out_path, override=True)

# Example 2: Load centroids from an existing file
centroids = calc_poi_labeled_buffered(msk_reference, subreg_reference, out_path, override=False)
```

## Code Analysis
### Inputs
- `msk_reference` (Image_Reference): The mask to compute the centroids from.
- `subreg_reference` (Image_Reference | None): The subregion mask to compute the centroids relative to. If None, only `msk_reference` is used.
- `out_path` (Path | str): The path to save the computed centroids to (required — automatic path generation is deprecated).
- `subreg_id` (int | Location | list[int | Location] | None, optional): The ID(s) of the subregion(s) to compute centroids in.
- `verbose` (bool, optional): Flag to print verbose output.
- `override` (bool, optional): Overwrite an existing centroids file at `out_path`.
- `decimals` (int, optional): Number of decimal places to round the computed centroid coordinates to. Default is 3.
- `check_every_point` (bool, optional): If True, re-computes centroids even when `out_path` already exists, so that every point is verified against the source segmentation. Default is True.
___
### Flow
If the output file exists and `override=False`, the function loads and returns the existing centroids. Otherwise, it computes the centroids using the given mask and subregion references, saves them to `out_path`, and returns them as a `POI` object.
___
### Outputs
The computed centroids, as a `POI` object.
___
